```python
import os
import re

def remove_comments(file_path):
    """Remove comments from a file."""
    with open(file_path, 'r', encoding='utf-8') as file:
        code = file.read()

    # Regular expressions to match single-line and multi-line comments
    single_line_comment = r'//.*'
    multi_line_comment = r'/\*.*?\*/'

    # Remove single-line and multi-line comments
    code_no_comments = re.sub(single_line_comment, '', code)
    code_no_comments = re.sub(multi_line_comment, '', code_no_comments, flags=re.DOTALL)

    # Write the cleaned code back to the file
    with open(file_path, 'w', encoding='utf-8') as file:
        file.write(code_no_comments)

def traverse_and_process(directory):
    """Traverse directory and process .c and .cpp files."""
    for root, _, files in os.walk(directory):
        for file in files:
            if file.endswith(('.c', '.cpp')):
                file_path = os.path.join(root, file)
                print(f"Processing file: {file_path}")
                remove_comments(file_path)

if __name__ == "__main__":
    target_directory = "testcodes"
    traverse_and_process(target_directory)
```
